```python
import numpy as np
import matplotlib.pyplot as plt
t = np.linspace(0, 40, 100)
d_r = 2.5*t
d_s = 3*(t-5)
fig, ax = plt.subplots()
plt.title("Distance vs Time")
plt.xlabel("Time (min)")
plt.ylabel("Distance (km)")
ax.set_xlim([0, 40])
ax.set_ylim([0, 100])
ax.plot(t, d_r, c='red', label="d_r")
ax.plot(t, d_s, c='blue', label="d_s")
plt.axvline(x=30, c='green', label="t=30s", linestyle="-.")
_ = plt.axhline(y=75, c='green', label="d=75km", linestyle="--")
```
